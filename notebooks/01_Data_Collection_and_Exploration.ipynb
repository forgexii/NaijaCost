## 2. Automated Web Scraping & Data Collection

We fetch the comprehensive **UN World Food Programme (WFP) Nigeria Food Prices Database** from the Humanitarian Data Exchange (HDX).
This dataset contains over 90,000 empirical price observations spanning from 2002 to present.

```python
#Import the required libraries
import os
import urllib.request
from pathlib import Path

# Create a directory to store the raw data files
raw_dir = Path("../data/raw")
# Create the directory if it doesn't exist
raw_dir.mkdir(parents=True, exist_ok=True)

# Define the file paths for the raw data files
prices_file = raw_dir / "wfp_food_prices_nga.csv"
markets_file = raw_dir / "wfp_markets_nga.csv"

# Define the URLs for the raw data files
urls = {
    prices_file: "https://data.humdata.org/dataset/42db041f-7aaf-4ab4-961f-2a12096861e7/resource/12b51155-0cd3-4806-9924-61ede4077591/download/wfp_food_prices_nga.csv",
    markets_file: "https://data.humdata.org/dataset/42db041f-7aaf-4ab4-961f-2a12096861e7/resource/5329e772-0b74-4f65-8cc0-37a0915cc7e4/download/wfp_markets_nga.csv"
}

# Download the raw data files if they don't exist
for path, url in urls.items():
    if not path.exists():
        print(f"Downloading {path.name}...")
        # Use a custom User-Agent header to avoid potential blocking by the server
        req = urllib.request.Request(url, headers={"User-Agent": "Mozilla/5.0"})
        #make the request and write the response to the file
        with urllib.request.urlopen(req) as resp, open(path, "wb") as f:
            f.write(resp.read())
    print(f"{path.name} verified at {path} ({path.stat().st_size / 1e6:.2f} MB)")
```

## 3. Loading the Granular Nigeria Market Dataset

```python
# Load the substantial food prices dataset
wfp_df = pd.read_csv(prices_file)
print("Dataset loaded successfully!")
print(f"Total rows: {wfp_df.shape[0]:,} | Total columns: {wfp_df.shape[1]}")
wfp_df.head()
```

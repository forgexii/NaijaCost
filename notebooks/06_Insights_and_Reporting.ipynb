# NaijaCost - Part 6: Insights, Policy Implications & Final Report

## Executive Summary
The **NaijaCost** project analyzed over 91,000 empirical market observations collected across 14 Nigerian states and 118 markets from 2002 to 2026. This comprehensive study evaluated food inflation dynamics, regional price friction, transport-driven cost pass-through, and the acute impact of the May 2023 macroeconomic reforms (petrol subsidy removal and exchange rate float).

---

## 1. Key Empirical Findings

### A. The May 2023 Policy Shock & Food Inflation
- Following the dual reforms in mid-2023, average prices of essential staple food items escalated by **150% to over 300%**.
- **Rice (local)** surged from a pre-subsidy baseline average of ~NGN 450/KG to peak above NGN 1,300/KG.
- **Garri (white)**, traditionally the most accessible safety-net carbohydrate in Nigeria, recorded unprecedented price acceleration exceeding NGN 850/KG.

### B. Fuel Pass-Through Elasticity
- Econometric regression analysis reveals an **$R^2$ of 0.869** between commercial fuel prices and staple grain prices.
- The estimated coefficient indicates that every **100 NGN increase in petrol prices** transmits an estimated **52.4 NGN increase** per kilogram of staple food, driven by domestic logistics and haulage costs.

### C. Regional Fragmentation & Supply Chain Friction
- Statistical hypothesis testing (Welch's t-test: $t = 16.48, p < 10^{-50}$; ANOVA: $F = 88.42, p < 10^{-53}$) confirms that Nigerian markets remain spatially fragmented.
- High haulage costs and interstate security bottlenecks prevent effective arbitrage between Northern food production centers and Southern coastal consumption hubs.

### D. Intermediary Markups
- Retail markups over wholesale bulk prices average **15% to 35%**, representing a substantial burden on low-income urban households that lack the liquidity to buy in bulk.


```python
import pandas as pd
import numpy as np
from pathlib import Path

proc_dir = Path("../data/processed")
summary_file = proc_dir / "naijacost_monthly_summary.csv"
clean_file = proc_dir / "naijacost_cleaned_food_prices.csv"

df = pd.read_csv(clean_file)
df['date'] = pd.to_datetime(df['date'])

# Generate final executive metrics table
key_items = ['Rice (imported)', 'Rice (local)', 'Gari (white)', 'Beans (white)', 'Maize (white)', 'Yam']
df_report = df[(df['commodity_clean'].isin(key_items)) & (df['base_unit'] == 'KG')].copy()
df_report['period'] = np.where(df_report['date'] < '2023-06-01', 'Pre-May 2023', 'Post-May 2023')

summary_table = df_report.groupby(['commodity_clean', 'period'])['norm_price'].agg(['mean', 'median', 'std']).unstack()
summary_table['Price Increase (%)'] = ((summary_table['mean']['Post-May 2023'] - summary_table['mean']['Pre-May 2023']) / summary_table['mean']['Pre-May 2023'] * 100).round(1)

print("Executive Summary: Staple Food Inflation Metrics (NGN / KG)")
display(summary_table)
```

## 2. Strategic Policy Recommendations

### 1. Dedicated Agro-Freight Corridors & Transport Subsidies
- **Finding**: Fuel price volatility transmits directly into staple food costs ($R^2 = 0.869$).
- **Recommendation**: Federal and state governments should establish dedicated agricultural logistics corridors (rail freight from Kano/Kaduna to Lagos/Port Harcourt and subsidized CNG-powered agricultural haulage fleets) to decouple food transport from petrol/diesel price shocks.

### 2. Strategic Grain Reserves & Counter-Cyclical Market Intervention
- **Finding**: Lean season prices (May–July) rise 15–25% above post-harvest levels.
- **Recommendation**: Deploy public-private grain silo reserves with transparent trigger-release mechanisms during the lean season to suppress artificial price spikes.

### 3. Interstate Border Toll & Revenue Harmonization
- **Finding**: Spatial price parity tests indicate market fragmentation across geopolitical zones.
- **Recommendation**: Eliminate multiple informal checkpoints and interstate tax collection points along major transit routes (e.g. Kano-Lagos, Benue-Enugu corridors), which add 10–20% to food haulage overheads.

### 4. Consumer Bulk-Purchasing Cooperatives
- **Finding**: Retail consumers pay a 15–35% markup over wholesale prices.
- **Recommendation**: Community and workplace food purchasing cooperatives should be supported to enable low-income households to access wholesale prices directly.
